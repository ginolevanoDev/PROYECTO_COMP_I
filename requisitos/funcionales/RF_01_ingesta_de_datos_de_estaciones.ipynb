#Cargamos los archivos!

```python
# Cargamos las lecturas meteorológicas y la información de las estaciones
import pandas as pd

lecturas = pd.read_json("aruba_readings_20260924.json")
estaciones = pd.read_json("aruba_stations.json")
```

#Relacionamos

```python
# Relacionamos cada lectura meteorológica con su estación.
# station_id de las lecturas corresponde con id de aruba_stations.json.

datos = lecturas.merge(
    estaciones,
    left_on="station_id",
    right_on="id",
    how="left"
)
```

```python
datos.head()
```

#Validación de la relación entre lecturas y estaciones

```python
print("Lecturas:", len(lecturas))
print("Estaciones:", len(estaciones))
print("Datos unidos:", len(datos))

print("Lecturas sin estación:", datos["name"].isna().sum())
```

```python
print("Estaciones distintas en lecturas:",
      lecturas["station_id"].nunique())

print("Estaciones disponibles:",
      estaciones["id"].nunique())
```

```python
print(
    "IDs duplicados en estaciones:",
    estaciones["id"].duplicated().sum()
)
```
